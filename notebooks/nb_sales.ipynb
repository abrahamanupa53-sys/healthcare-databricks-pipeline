```python
#creating dataframe
data = [(1, 2, 500.0, 300.0, "Laptop Bag", "Riya"),
        (2, 1, 1200.0, 900.0, "Headphones", "Kabir"),
        (3, 3, 550.0, 320.0, "Laptop Bag", "Kabir"),
        (4, 4, 150.0, 80.0, "Mouse", "Sneha"),
        (5, 1, 1300.0, 950.0, "Headphones", "Riya")]
cols= ["sales_id", "sales_quantity", "selling_price", "cost",
        "product_name", "customer_name"]

df_sales = spark.createDataFrame(data, cols)
display(df_sales)
```

revenue = quantity × selling_price


profit = revenue − (quantity × cost)

```python

from pyspark.sql.functions import col
df_sales = (df_sales
        .withColumn("revenue", col("sales_quantity")* col("selling_price"))
        .withColumn("profit", col("revenue")- (col("sales_quantity")*col("cost"))))
display(df_sales)

```

Q1: highest cost per product
Q2: customer who bought the most items

```python
from pyspark.sql.functions import max, sum
df_highcost = df_sales.groupBy("product_name").agg(max("cost").alias("highest_cost"))
display(df_highcost)
```

```python
df_mostitemboughtcust = (df_sales.groupBy("customer_name")
                   .agg(sum("sales_quantity").alias("total_items"))
                .orderBy(col("total_items").desc()))
display(df_mostitemboughtcust)
```

```python
df_sales.write.format("delta").mode("overwrite").saveAsTable("default.sales")
```
